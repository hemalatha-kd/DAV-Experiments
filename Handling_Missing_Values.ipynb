```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena', 'Arjun'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'], 
    'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']  
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)
```

```python
print("Missing values in each column:")
print(df.isnull())

print("\nNumber of missing values in each column:")
print(df.isnull().sum())
```

```python
df_mean=df.copy()

print("Mean Age:",df_mean['Age'].mean())
df_mean['Age']=df_mean['Age'].fillna(df_mean['Age'].mean())
print("\nAfter Mean Imputation:")
print(df_mean)
```

```python
df_median = df.copy()

print("Median Salary:", df_median['Salary'].median())

# Corrected imputation line (fixed column name typo and invalid assignment)
df_median['Salary'] = df_median['Salary'].fillna(df_median['Salary'].median())

print("\nAfter Median Imputation:")
print(df_median)
```

```python
df_ffill=df.copy()

print("Before Forward Fill:")
print(df_ffill)

df_ffill['Department']=df_ffill['Department'].ffill()

print("\nAfter Forward Fill:")
print(df_ffill)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

data_encoding = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}

df_encoding = pd.DataFrame(data_encoding)
encoder = LabelEncoder()

df_encoding['Department_Encoded'] = encoder.fit_transform(df_encoding['Department'])

print(df_encoding)

print("\nClasses:")
print(encoder.classes_)
```

```python
import pandas as pd

data_encoding = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}

df_onehot = pd.DataFrame(data_encoding)

# Fixed the variable name here to match what is being printed (or vice versa)
encoded_df = pd.get_dummies(df_onehot, columns=['Department'], dtype=int)

print("After One-Hot Encoding:")
print(encoded_df)
```

```python
final_df=df.copy()

final_df['Age']=final_df['Age'].fillna(final_df['Age'].mean())

final_df['Salary']=final_df['Salary'].fillna(final_df['Salary'].median())

final_df['Department']=final_df['Department'].ffill()

print("Final Dataset:")
print(final_df)

print("\nChecking missing values:")
print(final_df.isnull().sum())
```
